```python
#List Comprehension

servers = [{"name":"server1","status":"UP"},
           {"name":"server2","status":"DOWN"},
            {"name":"server3","status":"UP"}
           ]

names = [server["name"] for server in servers if  server["status"] == "UP"]
print (names)
```

```python
volumes =[
    {"name":"vol1","size":100},
    {"name":"vol2","size":200},
    {"name":"vol3","size":300}
]

large_volume = [v["name"] for v in volumes if v["size"] > 200]
print (large_volume)
```

```python
# Nested List Comprehension

matrix = [
    [1,2],
    [3,4],
    [5,6]]

result = [num for row in matrix for num in row]

print (result)
```

```python
#Given numbers = [10,15,20,25,30,35] extract only number divisible by 10

numbers = [10,15,20,25,30,35]

result = [i for i in numbers if i % 10 == 0 ]

print(result)
```

```python
# convert all the name to upper case
names = ['aniket', 'rahul', 'amit']

upper_case = [name.upper() for name in names ]

print( upper_case)
```

```python
# Create a list of numbers from 1 to 20 that are divisible by 3

list = []

list = [i for i in range(1,21) if i % 3 == 0]

print(list)
```

```python
# Given numbers =[1,2,3,4,5], create a list containing 'even' or 'odd' for each number
numbers =[1,2,3,4,5]

list = []

list = ['even' if i % 2 == 0 else 'odd' for i in numbers ]

print(list)
```
